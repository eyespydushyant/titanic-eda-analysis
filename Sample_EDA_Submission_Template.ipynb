## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
df1 = df.copy()
df1.isnull().sum()

````

````python
# Fill missing values in the Age column with the median age
df1['Age'].fillna(df1['Age'].median(), inplace=True)
````

````python
# Fill missing values in the Embarked column with the most frequent value
df1['Embarked'].fillna(df1['Embarked'].mode()[0], inplace=True)
````

````python
# Drop the Cabin column because it contains a large number of missing values
df1.drop('Cabin', axis=1, inplace=True)
````

````python
# Check for duplicate rows
print("Duplicate Rows:", df1.duplicated().sum())
````

````python
df1.isnull().sum()
````

### What all manipulations have you done and insights you found?

What all manipulations have you done and insights you found?

To make the dataset ready for analysis, several data wrangling steps were performed. First, a copy of the original dataset was created to preserve the raw data. Missing values were identified using the isnull() function. The missing values in the Age column were replaced with the median age because the median is less affected by outliers. The missing values in the Embarked column were filled with the mode (most frequent value), as it is a categorical variable. The Cabin column was dropped because it contained a very high number of missing values (687), making it less useful for analysis. The dataset was also checked for duplicate records, and no duplicate rows were found. Finally, the dataset was verified to ensure that no missing values remained in the columns used for analysis.

Insights Found
